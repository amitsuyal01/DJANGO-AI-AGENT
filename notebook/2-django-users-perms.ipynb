### get staff users

```python
User.objects.all()
```

```python
staff_user  = User.objects.filter(is_superuser=False, is_staff=True).first()
print(staff_user)
staff_user.user_permissions.all()
```

```python
#import the permission model
from django.contrib.auth.models import Permission
```

```python
perms= Permission.objects.filter(codename__in=['add_document', 'change_document', 'delete_document', 'view_document'])
for p in perms:
    staff_user.user_permissions.add(p)
```
