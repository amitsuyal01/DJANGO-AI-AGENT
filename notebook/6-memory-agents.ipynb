```python
from langgraph.checkpoint.memory import InMemorySaver
checkpoint_saver = InMemorySaver()
```

```python
from ai.agents import get_document_agent
```

```python
agent = get_document_agent(checkpointer=checkpoint_saver)
```

```python
import uuid

config = {"configurable": {"user_id": "2"}, "thread_id": str(uuid.uuid4())}

response = agent.invoke(
    {"messages": 
        [
            {"role": "user", "content": "what are the most recent documents?"}
         ]
    },
    config=config
    )
```

```python
for msg in response["messages"] :
    print (msg.content)
```

```python
response["messages"][-1].content
```
